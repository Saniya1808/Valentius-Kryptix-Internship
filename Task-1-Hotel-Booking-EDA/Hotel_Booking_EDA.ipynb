```python
# Missing value Treatment

# For Numerical use-> Mean or Median or KNNImputer method
# For Categorical use-> Mode Method

# Fill missing children values with the mode
df["children"] = df["children"].fillna(df["children"].mode()[0])

# Represent missing country information explicitly
df["country"] = df["country"].fillna("Unknown")

# Remove company because most of its values are missing
df = df.drop(columns=["company"])
```

```python
# Duplicates Value Treatment

duplicate_rows = df[df.duplicated(keep=False)]

print("Duplicate rows including all copies:", duplicate_rows.shape[0])
print("Unique duplicate rows:", duplicate_rows.drop_duplicates().shape[0])

duplicate_rows.head(10)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of exact duplicate rows:", duplicate_count)
print("Percentage of exact duplicate rows:",
      round((duplicate_count / len(df)) * 100, 2), "%")
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
print("Shape after removing duplicates:", df.shape)
print("Remaining exact duplicates:", df.duplicated().sum())
```

```python
# Otliers treatment
# Find IQR Find lw and uw
import numpy as np

def wisker(col):
    q1, q3 = np.percentile(col, [25, 75]) # Corrected typo here
    iqr = q3 - q1
    lw = q1 - 1.5 * iqr
    uw = q3 - 1.5 * iqr
    return lw, uw
```

```python
# Assuming df is your pandas DataFrame

numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

for i in numerical_cols:
    lw, uw = wisker(df[i])
    df[i] = np.where(df[i] < lw, lw, df[i])
    df[i] = np.where(df[i] > uw, uw, df[i])
```

```python
# Check Outliers are removed or not
for i in ['hotel', 'is_canceled', 'lead_time', 'arrival_date_year',
       'arrival_date_month', 'arrival_date_week_number',
       'arrival_date_day_of_month', 'stays_in_weekend_nights',
       'stays_in_week_nights', 'adults', 'children', 'babies', 'meal',
       'country', 'market_segment', 'distribution_channel',
       'is_repeated_guest', 'previous_cancellations',
       'previous_bookings_not_canceled', 'reserved_room_type',
       'assigned_room_type', 'booking_changes', 'deposit_type', 'agent',
       'days_in_waiting_list', 'customer_type', 'adr',
       'required_car_parking_spaces', 'total_of_special_requests',
       'reservation_status', 'reservation_status_date']:
    sns.boxplot(df[i])
    plt.show()
```

```python
# Garbage value traetment

#Check exactly which rows contain "Undefined"

for col in ["meal", "market_segment", "distribution_channel"]:
    print(f"\n{'='*50}")
    print(f"Column: {col}")
    print(df[df[col] == "Undefined"][col].value_counts())
    print(df[df[col] == "Undefined"].head())
```

```python
# Check rows containing suspicious negative/fractional values
suspicious_rows = df[
    (df["is_canceled"] < 0) |
    (df["lead_time"] < 0) |
    (df["arrival_date_week_number"] % 1 != 0) |
    (df["arrival_date_day_of_month"] % 1 != 0) |
    (df["stays_in_weekend_nights"] < 0) |
    (df["stays_in_week_nights"] < 0) |
    (df["total_of_special_requests"] < 0)
]

print("Number of suspicious rows:", len(suspicious_rows))
print("Percentage of suspicious rows:",
      round(len(suspicious_rows) / len(df) * 100, 2), "%")
```

## Visualization 1 — Which hotel type gets more bookings?

```python
sns.countplot(data=df, x="hotel")

plt.title("Number of Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")

plt.show()
```

### Interpretation:

City Hotel has significantly more bookings than Resort Hotel, indicating that City Hotel is more frequently booked in this dataset.

## Visualization 2 — Which month has the most bookings?

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

sns.countplot(
    data=df,
    x="arrival_date_month",
    order=month_order
)

plt.title("Number of Bookings by Arrival Month")
plt.xlabel("Arrival Month")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)

plt.show()
```

### Interpretation:

August has the highest number of bookings, while January has the lowest. This indicates that booking demand varies considerably across months.

## Visualization 3 — What is the distribution of lead time?

```python
sns.histplot(data=df, x="lead_time", bins=50, kde=True)

plt.title("Distribution of Booking Lead Time")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Number of Bookings")

plt.show()
```

### Interpretation:

Most bookings are made with relatively short lead times, while a smaller number of bookings are made much further in advance. The distribution is right-skewed.

## Visualization 4 — How does ADR vary by hotel type?
ADR = Average Daily Rate

```python

sns.boxplot(data=df, x="hotel", y="adr")

plt.title("Average Daily Rate (ADR) by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate (ADR)")

plt.show()
```

### Interpretation:

The distribution of ADR differs between City Hotel and Resort Hotel. The box plot also shows the presence of extreme ADR values, which indicates considerable variation in room pricing

## Visualization 5 — Is lead time related to ADR?

```python
sns.scatterplot(
    data=df,
    x="lead_time",
    y="adr",
    alpha=0.3
)

plt.title("Relationship Between Lead Time and Average Daily Rate")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Average Daily Rate (ADR)")

plt.show()
```

### Interpretation:

The scatter plot shows considerable variation in ADR across different lead times. There does not appear to be a strong simple linear relationship between lead time and ADR.

## Visualization 6: Correlation Heatmap

```python
plt.figure(figsize=(12, 8))
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")

plt.show()
```

### Interpretation:

The heatmap shows the strength and direction of relationships between numerical variables. Most variables have weak to moderate correlations, while a few pairs show stronger relationships that may require further investigation.
